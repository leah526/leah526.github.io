# Project 1: US Flight Arrival Delays


For this project, I use the US Flight Delays & Cancellations
2025–2026 dataset available on
[Kaggle](https://www.kaggle.com/datasets/samuelpombourcq/us-flight-delays-and-cancellations-2025-2026).

I use flights_sample.csv, which contains 500,000 flight records. Specifically, I focus on the `arr_delay` column, which measures how many minutes a flight arrives earlier or later than its scheduled arrival time.
I compare its mean, median, and mode using pandas and then Python's standard library.

To run this notebook, download `flights_sample.csv` from the Kaggle link above and place it in the same folder as this notebook.


## Step 1: Read the data

I load the flight sample into a pandas DataFrame and preview the first five rows to check that it loaded correctly.

```python
import pandas as pd

file_path = "flights_sample.csv"
flights = pd.read_csv(file_path)

flights.head()
```

## Step 2: Check the data

I check the total number of rows, the data type of the arrival-delay column, and the number of missing values to get an overview of the dataset.

```python
print("Total rows:", len(flights))
print("Column type:", flights["arr_delay"].dtype)
print("Missing arrival delays:", flights["arr_delay"].isna().sum())
```

The dataset contains 500,000 flight records. The arrival-delay column is stored as a numeric data type. There are 10,364 missing arrival-delay values, which I exclude from the calculations in the next step.

## Step 3: Data cleaning

In this step, I exclude missing values from the calculations. I then check the number of valid values remaining.

```python
arrival_delays = flights["arr_delay"].dropna()

print("Valid arrival delays:", len(arrival_delays))
print("Excluded missing values:", len(flights) - len(arrival_delays))
```

## Step 4: Computing the mean, median, and mode

I use pandas to calculate the mean, median, and mode of the arrival delays. 

```python
mean_delay = arrival_delays.mean()
median_delay = arrival_delays.median()
mode_delay = arrival_delays.mode()

print("Mean arrival delay (minutes):", mean_delay)
print("Median arrival delay (minutes):", median_delay)
print("Mode arrival delay(s) (minutes):", mode_delay.tolist())
```

The results show that flights in this sample arrived 8.83 minutes late on average. The median arrival delay was −6 minutes, indicating an early arrival of 6 minutes. In addition, the most common outcome was arriving 13 minutes early.

## Step 5: Calculating mean, median, and mode using the hard way

In this step, I calculate the mean, median, and mode using Python's standard library.

```python
import csv

file_path = "flights_sample.csv"
delays = []

with open(file_path, "r", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        value = row["arr_delay"]

        if value != "":
            delays.append(float(value))

print("Valid arrival delays:", len(delays))
```

After excluding missing arrival-delay values, the number of valid records is 489,636, which is the same as the number obtained using pandas.

```python
total_delay = 0

for delay in delays:
    total_delay = total_delay + delay

mean_delay_hard = total_delay / len(delays)

print("New mean arrival delay (minutes):", mean_delay_hard)
```

```python
sorted_delays = sorted(delays)
n = len(sorted_delays)
middle = n // 2

if n % 2 == 0:
    median_delay_hard = (sorted_delays[middle - 1] + sorted_delays[middle]) / 2
else:
    median_delay_hard = sorted_delays[middle]

print("New median arrival delay (minutes):", median_delay_hard)
```

```python
delay_counts = {}

for delay in delays:
    if delay in delay_counts:
        delay_counts[delay] = delay_counts[delay] + 1
    else:
        delay_counts[delay] = 1

highest_count = max(delay_counts.values())

mode_delay_hard = []

for delay in delay_counts:
    if delay_counts[delay] == highest_count:
        mode_delay_hard.append(delay)

mode_delay_hard.sort()

print("New mode arrival delay(s) (minutes):", mode_delay_hard)
```

The mean, median, and mode calculated using Python’s standard library match the results obtained using pandas.

## Step 6: Visualizing arrival delays

I group flights into arrival-delay categories and use a sparkline-style bar chart to show their distribution. Longer bars represent more flights.

```python
categories = {
    "Early": 0,
    "On time": 0,
    "1–15 min late": 0,
    "16–60 min late": 0,
    "Over 60 min late": 0,
}

for delay in delays:
    if delay < 0:
        categories["Early"] += 1
    elif delay == 0:
        categories["On time"] += 1
    elif delay <= 15:
        categories["1–15 min late"] += 1
    elif delay <= 60:
        categories["16–60 min late"] += 1
    else:
        categories["Over 60 min late"] += 1

largest_count = max(categories.values())

print("Arrival-delay distribution")
print("Bar lengths show flight counts; the longest bar has 40 stars.")
print()

for label, count in categories.items():
    bar_length = round(count / largest_count * 40)
    bar = "*" * bar_length
    percentage = count / len(delays) * 100

    print(f"{label:<17} | {bar:<40} {percentage:5.1f}%")
```

The chart shows that 60.4% of flights with valid arrival-delay data arrived early, while 1.7% arrived exactly on time. The remaining 37.9% arrived late, but most delayed flights arrived no more than an hour after their scheduled arrival time.

These results describe only the sampled flights with non-missing arrival-delay data.
